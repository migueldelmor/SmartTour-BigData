```python
import itertools as it
import pandas as pd

camisetas = ["C1","C2","C3"]
pantalones = ["P1","P2"]
zapatos = ["Z1","Z2"]

#Calcula cuántos outfits diferentes se pueden formar en total utilizando el principio multiplicativo.

outfitsDiferentes = len(camisetas)*len(pantalones)*len(zapatos)

print(outfitsDiferentes)

#combinaciones
cartesiano = list(it.product(camisetas,pantalones,zapatos))

df = pd.DataFrame(cartesiano, columns=['camisetas','pantalones','zapatos'])

print(df)
```
