```python
import os

import pandas as pd


jugadoresA = set()
jugadoresB = set()

for archivo in os.listdir("../datasets/equipoA"):
    ruta = os.path.join("../datasets/equipoA", archivo)
    dfjugadoresA = pd.read_csv(ruta)
    jugadoresA.update(dfjugadoresA["jugador"].dropna().astype(str))


for archivo in os.listdir("../datasets/equipoB"):
    ruta = os.path.join("../datasets/equipoB", archivo)
    dfjugadoresB = pd.read_csv(ruta)
    jugadoresB.update(dfjugadoresB["jugador"].dropna().astype(str))

#Muestra los conjuntos ordenados alfabeticamente

print(jugadoresA)
print(jugadoresB)

#1. Calcula la **union** de ambos conjuntos.- Cuantos jugadores diferentes han pasado por alguno de los dos clubes?

union = jugadoresA | jugadoresB

print(union)

#2. Calcula la **interseccion**.- Que jugadores han pertenecido a ambos clubes?- Cuantos son?

interseccion = jugadoresA & jugadoresB

print(interseccion)

#3. Calcula la **diferencia** entre el Club A y el Club B.- Que jugadores han jugado unicamente en el Club A?

diferenciaA = jugadoresA - jugadoresB
diferenciaB = jugadoresB - jugadoresA

print(diferenciaA)
print(diferenciaB)

#4. Calcula la **diferencia simétrica**.- ¿Qué jugadores han jugado exclusivamente en uno de los dos clubes?

diferenciaSimetricaA = jugadoresA ^ jugadoresB
diferenciaSimetricaB = jugadoresB ^ jugadoresA

#5. Considerando como **universo** la unión de ambos conjuntos, calcula el **complemento** de la intersección.- ¿Qué jugadores no han pasado por ambos clubes?

complemento = union - interseccion

print(complemento)

#Parte 3. Análisis de resultados

#1. ¿Qué porcentaje de jugadores del universo ha jugado en ambos clubes?

porcentaje = (len(interseccion) / len(union)) * 100

print(porcentaje)

#2. ¿Qué club ha tenido un mayor número de jugadores exclusivos?

if(len(diferenciaA) > len(diferenciaB)):
    print("Equipo A tiene mas jugadores exclusivos")
elif (len(diferenciaA) > len(diferenciaB)):
    print("Equipo B tiene mas jugadores exclusivos")
else:
    print("Los dos tienen el mismo numero de jugadores exclusivos")
```
