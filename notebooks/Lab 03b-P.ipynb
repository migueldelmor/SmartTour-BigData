```python
import pandas as pd

ventas = pd.read_excel("../datasets/ventas.xlsx")

clientes = pd.read_csv("../datasets/clientes.csv")

contactos = pd.read_json("../datasets/contacto.json")

#Quitamos duplicados
clientes = clientes.drop_duplicates()
ventas = ventas.drop_duplicates()
contactos = contactos.drop_duplicates()

#Normalizamos nombres, quitamos espacios y ponemos nombre propio
clientes["nombre"] = clientes["nombre"].str.strip().str.capitalize()

#Hacemos lo mismo con las ciudades, con title para que ponga ok el nombre de las ciudades
clientes["ciudad"] = clientes["ciudad"].str.strip().str.title()

#Estandarización de fechas

clientes["fecha_alta"] = pd.to_datetime(clientes["fecha_alta"], format="%d/%m/%Y", errors="coerce").dt.strftime("%Y-%m-%d")

#Renombramos la columna de clientes para poder hacer el merge

clientes = clientes.rename(columns={"id_cliente" : "cliente_id"})

#ahora podemos mergear
clientes_ventas = pd.merge(clientes,ventas,on="cliente_id", how="left")
clientes_ventas_contacto = pd.merge(clientes_ventas,contactos, on="cliente_id", how="left")

clientes_ventas_contacto.to_csv("../datasets/resultado.csv", index=False, encoding="utf-8", lineterminator="\n")


```
