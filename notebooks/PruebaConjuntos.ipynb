```python
import csv

grupo1 = set(['Alumno1','Alumno2','Alumno3','Alumno5'])
grupo2 = set(['Alumno1','Alumno3','Alumno4','Alumno5'])

union = grupo1 | grupo2
interseccion = grupo1 & grupo2
diferencia = grupo1 - grupo2
diferenciaSim = grupo1 ^ grupo2


print(union)
print(interseccion)
print(diferencia)
print(diferenciaSim)


with open('../datasets/grupo1.csv', 'w', newline='') as csvfile:
    writer = csv.writer(csvfile, delimiter=',',
                            quotechar='|', quoting=csv.QUOTE_MINIMAL)
    writer.writerow(grupo1)

with open('../datasets/grupo2.csv', 'w', newline='') as csvfile:
    writer = csv.writer(csvfile, delimiter=',',
                            quotechar='|', quoting=csv.QUOTE_MINIMAL)
    writer.writerow(grupo2)
```
