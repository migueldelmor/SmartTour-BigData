```python
import pandas as pd

ventas = pd.read_csv("../datasets/ventas.csv")

productos = pd.read_excel("../datasets/productos.xlsx")

opiniones = pd.read_json("../datasets/opiniones.json")

#Valores nulos en ventas
print(ventas.isnull().sum())

#Valores nulos en productos
print(productos.isnull().sum())

#Valores nulos en opiniones
print(opiniones.isnull().sum())

#Valores duplicados en ventas
print(ventas.duplicated().sum)

#Valores duplicados en productos
print(productos.duplicated().sum)

#Valores duplicados en opiniones
print(opiniones.duplicated().sum)

ventas_productos = pd.merge(ventas, productos, on="producto_id", how="left")

ventas_productos_opiniones = pd.merge(ventas_productos, opiniones, on="cliente_id",  how="left")

#Producto mas vendido
#print(ventas_productos_opiniones["nombre"].value_counts()) #cuanto se ha vendido de cada producto
#print(ventas_productos_opiniones["nombre"].max()) #producto mas vendido


#Total Gastado por cliente
ventas_productos_opiniones["total"] = ventas_productos_opiniones["cantidad"] * ventas_productos_opiniones["precio"]
gasto_cliente = ventas_productos_opiniones.groupby("cliente_id")["total"].sum()
#print(gasto_cliente)

#Cliente que mas ha gastado
maxCliente = gasto_cliente[gasto_cliente == gasto_cliente.max()]
for cliente_id, total in maxCliente.items():
    print(f"El cliente con mas gasto es {cliente_id} con un gasto de {total}")

#Valoracion media por cliente
valoraciones = ventas_productos_opiniones.groupby("cliente_id")["valoracion"].mean()
print(valoraciones)



```
